```python
df = pd.read_csv("./data/data2025.csv")
df.head()
```

```python
# 환적만 선택
trans = df[df["구분"] == "환적"]

# 연도별 TOP 10
trans_top10 = (
    trans.sort_values(["연도", "계"],
    ascending=[True, False])
    .groupby("연도")
    .head(10)
)

# 2025년 TOP 10만 선택
trans_top10_2025 = trans_top10[trans_top10["연도"] == 2025]

fig_trans_top10_2025 = px.bar(
    trans_top10_2025,
    x="국가명",
    y="계",
    title="2025년 부산항 환적 물동량 TOP 10 국가"
)

fig.show(renderer="browser")
```

```python
# 입항만 선택
arrival = df[df["구분"] == "입항"]

# 연도별 TOP 10
arrival_top10 = (
    arrival.sort_values(["연도", "계"],
    ascending=[True, False]).groupby("연도").head(10)
)

# 2025년 TOP 10만 선택
arrival_top10_2025 = arrival_top10[arrival_top10["연도"] == 2025]

fig_arrival_top10_2025 = px.bar(
    arrival_top10_2025,
    x="국가명",
    y="계",
    title="2025년 부산항 입항 물동량 TOP 10 국가"
)

fig_arrival_top10_2025.show(renderer="browser")
```

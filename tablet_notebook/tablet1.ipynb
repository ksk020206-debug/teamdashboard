```python
# 물동량 데이터를 숫자형으로 변환
df1['2020'] = pd.to_numeric(df1['2020'], errors='coerce')
df1['2021'] = pd.to_numeric(df1['2021'], errors='coerce')df1.isna().sum()
```

```python
# 물동량이 없는 빈칸을 0으로 처리
df1[['2020', '2021']] = df1[['2020', '2021']].fillna(0)

# 결측치가 모두 처리됐는지 확인
df1.isna().sum()
```

```python
# 2020, 2021 연도별 열을 세로 형태로 변환
df1_long = df1.melt(
    id_vars=['입출항구분', '국가명'],
    value_vars=['2020', '2021'],
    var_name='연도',
    value_name='물동량'
)

df1_long.head()
```

```python
df1_long['연도'] = df1_long['연도'].astype(int)
```

```python
df1_long.dtypes
```

```python
df1_long.shape
```

```python
df1_long['입출항구분'].value_counts()
```

```python
# 수입 데이터
import_df = df1_long[
    df1_long['입출항구분'] == '입항'
].copy()

import_df.head()
```

```python
# 수출 데이터
export_df = df1_long[
    df1_long['입출항구분'] == '출항'
].copy()

export_df.head()
```

```python
# 환적 데이터만 선택
trans_df = df1_long[
    df1_long['입출항구분'].isin(['입항환적', '출항환적'])
].copy()

trans_df.head()
```

```python
# 입항환적 + 출항환적 합계
trans_df = (
    trans_df
    .groupby(['연도', '국가명'])['물동량']
    .sum()
    .reset_index()
)

trans_df.head()
```

```python
# 국가별 전체 물동량
total_df = (
    df1_long
    .groupby(['연도', '국가명'])['물동량']
    .sum()
    .reset_index()
)

total_df.head()
```
